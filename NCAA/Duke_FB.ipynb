```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re
```

```python
url = "https://goduke.com/sports/football/roster"

# Set headers to mimic a browser request
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

# Fetch the page
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.content, 'html.parser')
```

```python
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.content, 'html.parser')

# Initialize list to store player data
players_data = []

# Find all player cards
player_cards = soup.find_all('div', class_='s-person-card')

for card in player_cards:
    # Extract name from the h3 tag
    name_elem = card.find('h3')
    if name_elem:
        full_name = name_elem.get_text(strip=True)
    else:
        continue
    
    # Extract position - look for text that says "Position" followed by the position abbreviation
    # The structure shows "Position CB" or "Position S" etc.
    position_text = card.get_text()
    
    # Find "Position" in the text and get what follows it
    position = ''
    if 'Position' in position_text:
        # Split by "Position" and get the next word
        parts = position_text.split('Position')
        if len(parts) > 1:
            # Get the text after "Position" and extract first word (the position abbreviation)
            after_position = parts[1].strip().split()[0]
            position = after_position
    
    # Split name into first and last
    name_parts = full_name.split()
    first_name = name_parts[0] if name_parts else ''
    last_name = ' '.join(name_parts[1:]) if len(name_parts) > 1 else ''
    
    players_data.append({
        'First Name': first_name,
        'Last Name': last_name,
        'Position': position
    })
```

```python
# Create DataFrame
duke = pd.DataFrame(players_data)

# Display the results
print(f"Total players scraped: {len(duke)}")
duke
```
